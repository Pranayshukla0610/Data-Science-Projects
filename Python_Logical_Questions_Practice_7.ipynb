```python
#Find the Second Largest Number in a List

numbers = [12, 45, 7, 34, 89, 23, 89, 67]

largest = second_lar = 0

for num in numbers:
  if num > largest:
    second_lar = largest
    largest = num
  elif num > second_lar and num != largest:
    second_lar = num

print(second_lar)
```

```python
#Count Frequency of Every Element

numbers = [1,2,2,3,3,3,4,4,5]

freq = {}

for num in numbers:
  if num in freq:
    freq[num] += 1
  else:
    freq[num] = 1

print(freq)
```

```python
#Check Whether Two Strings are Anagrams

str1 = input("Enter first string:")
str2 = input("Enter second string:")

if sorted(str1) == sorted(str2):
  print("Anagram String")
else:
  print("Not Anagram String")
```

```python
#Find Common Elements Between Two Lists

list1 = [1,2,3,4,5]

list2 = [3,4,5,6,7]

common = []

for num in list1:
  if num in list2:
    common.append(num)

print(common)
```

```python
#Find the Longest Word in a Sentence

sentence = "Python programming is very interesting"

words = sentence.split()

longest = words[0]

for word in words:
  if len(word) > len(longest):
    longest = word

print(longest)
```

```python
#Count Vowels and Consonants

str = input("Enter a string").lower()

vow = 0
cons = 0

for char in str:
  if char in "aeiou":
    vow += 1
  elif char.isalpha():
    cons += 1

print("Vowels:",vow)
print("Consonants:",cons)
```

```python
#Find All Prime Numbers Between 1 and 100

for num in range(2,101):

    prime = True

    for i in range(2,int(num**0.5)+1):

        if num % i == 0:
            prime = False
            break

    if prime:
        print(num,end=" ")
```
